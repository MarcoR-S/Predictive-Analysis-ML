1. **Bibliotecas e Dados**

```python
#Importando bibliotecas e funções
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

#Importando e Lendo o documento csv. Salvamos ele como um dataFrame
df = pd.read_csv("/content/fertilizer_prediction.csv") #Pode ser qualquer outro CSV
df.head(3) #Exibe as 3 primeiras linhas do dataframe
```

2. **Análise exploratória**

```python
#Informações sobre o csv
df.info()
df.describe()
#Gráfico para observar a quantidade da variável y da nossa análise (queremos prever)
sns.countplot(x='Fertilizer Name', data=df, color = "red", edgecolor="black")
plt.title("FERTILIZANTES")
plt.xticks(rotation=45)
```

3. **Limpeza de Dados**

```python
#Verificar a existência de duplicatas
duplicatas = df.duplicated().sum()
print("D: ", duplicatas)
#df = df.drop_duplicates(): caso existam


#Verificar a existência de outliers
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.show()
```

# 4. **Tradução Para a Linguagem do Computador**







```python
x = df.drop("Fertilizer Name", axis=1) #Df sem a variável que desejamos prever
y = df['Fertilizer Name'] #Df apenas com a variável que desejamos prever

le = LabelEncoder() #LABEL ENCODER
y = le.fit_transform(y)

cols = ['Crop Type', 'Soil Type'] #Colunas que queremos encodar
ohe = OneHotEncoder(handle_unknown='ignore')
x_encoded = pd.DataFrame(ohe.fit_transform(x[cols]).toarray())
x_encoded = x_encoded.add_prefix("OHE")
x = x.drop(cols, axis = 1) #Remove colunas com valores nominais
x = pd.concat([x, x_encoded], axis = 1) #Inclui coluinas com valores legiveis para o algoritmo

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) #Cria variaveis de teste e treino

scaler = MinMaxScaler() #Cria uma escala para padronizar os dados
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

* Label Encoder:  utilizado para a coluna que desejamos prever ou para dados com ordem lógica (Ordinais)

* One Hot Encoder:  utilizado para Dados Nominais

* Scaler: Padroniza a unidade de medida, a fim de manter apenas a proporção entre os dados
  1.   MinMaxScaler: Define o valor mínimo dos dados como 0 e o máximo como 1. Os outros dados são, proporcionalmente, igualados a um valor entre 0 e 1




5. **Criação de Modelos Preditivos**

```python
#REGRESSAO LOGISTICA
reglog = LogisticRegression(max_iter = 200)
reglog.fit(x_train_scaled, y_train)
predictReg = reglog.predict(x_test_scaled)
print("Acurácia REG: ", accuracy_score(y_test, predictReg))
print("Precisão: ", classification_report(y_test, predictReg))
AcuraciaREG = accuracy_score(y_test, predictReg)
#KNN
knn = KNeighborsClassifier(n_neighbors = 10)
knn.fit(x_train_scaled, y_train)
predictKnn = knn.predict(x_test_scaled)
print("Acurácia KNN: ", accuracy_score(y_test, predictKnn))
print("Precisão: ", classification_report(y_test, predictKnn))
AcuraciaKNN = accuracy_score(y_test, predictKnn)
#SVM

svmLin = SVC(kernel = 'linear')
svmLin.fit(x_train_scaled, y_train)
predictSVL = svmLin.predict(x_test_scaled)
print("Acurácia SVMLIN: ", accuracy_score(y_test,predictSVL))
print("Precisão: ",classification_report (y_test,predictSVL))
AcuraciaSVMLIN = accuracy_score(y_test,predictSVL)

svmPol = SVC(kernel ='poly')
svmPol.fit(x_train_scaled, y_train)
predictSVP = svmPol.predict(x_test_scaled)
print("Acurácia SVMPOL: ", accuracy_score(y_test, predictSVP))
print("Precisão: ", classification_report(y_test, predictSVP))
AcuraciaSVMPOL = accuracy_score(y_test, predictSVP)

svmRBF = SVC(kernel ='rbf')
svmRBF.fit(x_train_scaled, y_train)
predictSVR = svmRBF.predict(x_test_scaled)
print("Acurácia SVMRBF: ", accuracy_score(y_test, predictSVR))
print("Precisão: ", classification_report(y_test, predictSVR))
AcuraciaSVMRBF = accuracy_score(y_test, predictSVR)

#TREE
dt = DecisionTreeClassifier()
dt.fit(x_train_scaled, y_train)
predictDt = dt.predict(x_test_scaled)
print("Acurácia TREE: ", accuracy_score(y_test,predictDt))
print("Precisão: ",classification_report (y_test,predictDt))
AcuraciaTREE = accuracy_score(y_test,predictDt)

#FOREST
rf = RandomForestClassifier(n_estimators = 25)
rf.fit(x_train_scaled, y_train)
predictRf = rf.predict(x_test_scaled)
print("Acurácia RF: ", accuracy_score(y_test, predictRf))
print("Precisão: ", classification_report(y_test, predictRf))
AcuraciaFOREST = accuracy_score(y_test, predictRf)

```

6. **Exibir Resultados**

```python
resultado = {}
resultado["Regressão Logística "] = AcuraciaREG
resultado["KNN "] = AcuraciaKNN
resultado["SVM Linear "] = AcuraciaSVMLIN
resultado["SVM Polinomial "] = AcuraciaSVMPOL
resultado["SVM RBF "] = AcuraciaSVMRBF
resultado["Decision Tree "] = AcuraciaTREE
resultado["Random Forest "] = AcuraciaFOREST

df_ranking = pd.DataFrame(list(resultado.items()), columns =["Modelo", "Acurácia"])
df_ranking = df_ranking.sort_values(by="Acurácia", ascending = False)

print(df_ranking)
df_ranking.head(3)
```
